1) Cargar archivo

```python
import pandas as pd
#1. Cargar archivo CSV a un DataFrame

df = pd.read_csv('datos_practica_pandas.csv')

#2. Ver las primeras filas para tener una vista previa

print("---Vista Previa (HEAD)---")
print(df.head())

#3. Ver la cantidad de filas y columnas
print("---Cantidad de Filas y Columnas---")
print(df.shape)

```

2. Diagnostico tecnico

```python
print("--- Informacion general del dataframe")
df.info()
```

3. Eliminacion de filas vacias y de basura

```python

#1. Ver cuantas filas nulas hay antes de tocar nada
print("---Cantidad de Filas Nulas---")
print(df.isnull().sum())

#2. Eliminar las filas que esten completamente vacias
df.dropna(how='all', inplace=True)

#3. Eliminar filas si estuvieran totalmente duplicadas
df.drop_duplicates(inplace=True)

#Comparamos filas despues.
filas_despues = df.shape[0]
print("filas : ", filas_despues)
```

4.Limieza de textos y normalizacion de valores

```python
#1. Limpiar espacios extra o raros en la columna Average_Pulse
#Usamos .astype(str) para asegurar de que se lo trate como texto

df['Average_Pulse'] = df['Average_Pulse'].astype(str).str.replace(' ', '')

#2. Covertir columnas a números reales(float o int)
#errors=coerce encuentra texto y sino ppuede convertirlo en numero(AF) lo transforma
#en NaN

df['Average_Pulse']= pd.to_numeric(df['Average_Pulse'], errors='coerce')
df['Max_Pulse'] = pd.to_numeric(df['Max_Pulse'], errors='coerce')

#3 verificamos que ahora si sean columnas numericas
print("\n--- TIPOS DE DATOS ACTUALIZADOS ---")
print(df.dtypes)

```

5. Tratamiento de valores faltantes( IMPUTACIÓN O RELLENO)

```python
#1. Calcular el promedio de la columna Average_pulse (ignorando los NaN automaticamente)

promedio_pulso = df['Average_Pulse'].mean()
print("\n--- PROMEDIO DE PULSO ---")
print(promedio_pulso)

#2. Rellenar los valores nulos (Nan) de esa columna conn su respectivo promedio

df['Average_Pulse'] = df ['Average_Pulse'].fillna(promedio_pulso)

#3. hacer lo mismo para calorie_Burnage si tiene nulos
promedio_calorias = df['Calorie_Burnage'].mean()
df['Calorie_Burnage'] = df['Calorie_Burnage'].fillna(promedio_calorias)

#4. Rellenar los nulos de Max_pulse con su promedio
promedio_max_pulso = df['Max_Pulse'].mean()
df['Max_Pulse'] = df['Max_Pulse'].fillna(promedio_max_pulso)

#5. Rellenar los nulos Hours_Work con su promedio
promedio_horas_trabajo = df['Hours_Work'].mean()
df['Hours_Work'] = df['Hours_Work'].fillna(promedio_horas_trabajo)

#6. Verificamos que ya no queden nulos en esas columnas
print("\n--- CANTIDAD DE VALORES NULOS ---")
print(df.isnull().sum())


```

6. Analisis y reporte rápido


```python
#1. Ver un resumen estadistico completo de todo el dataset limpio
print("--- Resumen Estadistico---")
print(df.describe())
```

7. Se corrije la pulsación 9000 que aparece y es un error de tipeo

```python
#1. Detectar los valores absurdos en Averange_pulse ( por ejemplo , mayores a 250)
# y transformarlo en NaN( valor faltante)
import numpy as np


df.loc[df['Average_Pulse']> 250, 'Average_Pulse'] = np.nan

#2 Vemos como quedó la columna(ahora el 9000 a ser NaN)
print("--- Control de nulos despues de aislar el error---")
print(df['Average_Pulse'].isnull().sum())

#3 Calcular el Nuevo promedio real(sin el 9000)
nuevo_promedio = df['Average_Pulse'].mean()
print("\n--- NUEVO PROMEDIO DE PULSO ---")
print(nuevo_promedio)

#4. Rellenamos ese nuevo NaN con el promedio correcto
df['Average_Pulse']= df['Average_Pulse'].fillna(nuevo_promedio)

#5. controlamos de nuevo lo estadistico
print(df.describe())
```

7. Guardar y descargar el Csv limpio


```python
#1. Guardar el DataFrame limpio en un nuevo archivo
#usamos index= False ara que Pandas No guarde la columna de numeros fijos(0,1,2,3...) como coolumna extra

df.to_csv('datos_practica_pandas_limpios.csv', index=False)

#2. Descargar el archivo
from google.colab import files
files.download('datos_practica_pandas_limpios.csv')
```
