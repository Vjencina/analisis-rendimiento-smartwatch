Grafico de dispersion y Grafico de distribución

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargamos tu archivo limpio
df = pd.read_csv("datos_practica_pandas_limpios.csv")

# Ver las primeras filas para chequear que cargó bien
print(df.head())

# 2. Configuramos el estilo estético de los gráficos
sns.set_theme(style="whitegrid")

# --- GRÁFICO 1: Dispersión (Ej: Pulso vs Calorías) ---
# Cambia 'Calorie_Burnage' y 'Average_Pulse' por los nombres exactos de tus columnas
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='Average_Pulse', y='Calorie_Burnage', color='blue', alpha=0.7)

plt.title('Relación entre Pulso Promedio y Calorías Quemadas', fontsize=14)
plt.xlabel('Pulso Promedio', fontsize=12)
plt.ylabel('Calorías Quemadas', fontsize=12)
plt.show()

# --- GRÁFICO 2: Línea de tendencia temporal (Ej: Duración o Calorías por registro) ---
plt.figure(figsize=(10, 5))
sns.lineplot(data=df, y='Calorie_Burnage', x=df.index, color='green', linewidth=2)

plt.title('Comportamiento de las Calorías a lo largo de los registros', fontsize=14)
plt.xlabel('Número de Registro / Fila', fontsize=12)
plt.ylabel('Calorías Quemadas', fontsize=12)
plt.show()
```

### 📊 Conclusiones del Análisis de Rendimiento Físico

1. **Relación entre Esfuerzo y Gasto Calórico:**
   - Se observa una correlación positiva y directa entre el pulso promedio y las calorías quemadas.
   - A mayor exigencia cardíaca durante la sesión (pulso elevado), se registra un incremento proporcional en el gasto energético, validando que la intensidad impulsa la quema calórica.

2. **Comportamiento de la Carga de Entrenamiento:**
   - El gráfico temporal muestra una tendencia oscilante con picos y valles a lo largo de los registros.
   - Esto refleja la variabilidad natural de la rutina humana: los picos corresponden a días de alta intensidad o mayor duración, y los valles a jornadas de recuperación.

### 🏢 Aplicación de Negocio y Control de Calidad
* **Control de Calidad (Sanity Check):** La correlación positiva entre pulso y calorías valida que el algoritmo del dispositivo funciona coherentemente.
* **Valor para Empresas (Salud / Fitness):** Estos patrones permiten a las aplicaciones de bienestar personalizar alertas de entrenamiento y detectar variaciones de rendimiento en los usuarios.
